```python
import requests
import pandas as pd 
 
```

```python
url = 'https://api.mfapi.in/mf/125497'

response = requests.get(url, timeout=30)
response.raise_for_status()

print(response.status_code)
```

```python
result = response.json()

print(type(result))
print(result.keys())
```

```python
print(result['meta'])
```

```python
print(len(result['data']))
print(result['data'][:3])
```

```python
df = pd.DataFrame(result['data'])

print(df.shape)
print(df.size)
print(df.dtypes)
print(df.head())
```

```python
df.to_csv('../data/raw/nav_125497_raw.csv', index=False)
```
