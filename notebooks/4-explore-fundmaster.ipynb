```python
files = sorted(glob.glob('../data/raw/*.csv'))
datasets = {}

for f in files:
    name = os.path.basename(f).replace('.csv', '')

    if name.startswith('nav_'):
        continue

    short_name = name.split('-', 2)[-1]

    try:
        datasets[short_name] = pd.read_csv(f)
    except UnicodeDecodeError:
        datasets[short_name] = pd.read_csv(f, encoding='ISO-8859-1')

print(len(datasets))
print(list(datasets.keys()))
```

```python
master = datasets['01_fund_master']

print(master.shape)
print(master.dtypes)
master.head()
```

```python
CODE_COL = 'amfi_code'
NAME_COL = 'scheme_name'
FUND_HOUSE_COL = 'fund_house'
CATEGORY_COL = 'category'
SUBCATEGORY_COL = 'sub_category'
RISK_COL = 'risk_category'
```

```python
print('Number of fund houses:', master[FUND_HOUSE_COL].nunique())
print()
print(sorted(master[FUND_HOUSE_COL].dropna().unique()))
print()
print(master[FUND_HOUSE_COL].value_counts(dropna=False))
```

```python
columns_to_explore = {
    'Category': CATEGORY_COL,
    'Sub-category': SUBCATEGORY_COL,
    'Risk grade': RISK_COL,
}

for title, col in columns_to_explore.items():
    print('=' * 60)
    print(title, '| column:', col)
    print('=' * 60)
    print('Unique values:', master[col].nunique())
    print('Missing values:', master[col].isnull().sum())
    print()
    print(sorted(master[col].dropna().unique()))
    print()
    print(master[col].value_counts(dropna=False))
    print()
```

```python
master.groupby(CATEGORY_COL)[SUBCATEGORY_COL].unique()
```

```python
pd.crosstab(master[CATEGORY_COL], master[RISK_COL])
```

```python
print('Data type:', master[CODE_COL].dtype)
print('Smallest code:', master[CODE_COL].min())
print('Largest code:', master[CODE_COL].max())
print('Rows:', len(master))
print('Unique codes:', master[CODE_COL].nunique())
print()
print(master[CODE_COL].dropna().astype(int).astype(str).str.len().value_counts())
```

```python
by_code = master.sort_values(CODE_COL)

print(by_code[[CODE_COL, NAME_COL]].head(10))
print()
print(by_code[[CODE_COL, NAME_COL]].tail(10))
```
